```python
import cv2

# Read the image file
imgs = cv2.imread('UCA-7.jpg')  # ✅ CORRECT - reads the image

# Check if image was loaded successfully
if imgs is None:
    print("Error: Could not read the image file. Check if 'UCA-7.jpg' exists.")
    exit()

print(type(imgs))   # <class 'numpy.ndarray'>
print(imgs.shape)   # (height, width, channels) e.g., (480, 640, 3)
print(imgs)         # Prints the pixel values (large array)
```

```python
import cv2

# 1. READ the image
imgs = cv2.imread('UCA-7.jpg')

# 2. Check if image loaded properly
if imgs is None:
    print("❌ Image not found! Please check the file path.")
    exit()
else:
    print(f"✅ Image loaded successfully!")
    print(f"📐 Dimensions: {imgs.shape}")
    print(f"📊 Data type: {imgs.dtype}")
    print(f"🔢 Pixel values (first 10 pixels): {imgs[0][:10]}")

# 3. DISPLAY the image
cv2.imshow('UCA-7.jpg', imgs)

# 4. Wait for any key press
cv2.waitKey(0)

# 5. Clean up
cv2.destroyAllWindows()
```
